```python
#Remove punctuations, links, stopwords, mentions and \r\n new line characters
def cleaningText(text):
  if isinstance(text, str):
    text=re.sub(r'@[A-Za-z0-9]+', '', text)
    text=re.sub(r'#[A-Za-z0-9]+', '', text)
    text=re.sub(r'RT[\s]', '', text)
    text=re.sub(r"http\S+", '', text)
    text=re.sub(r'[0-9]', '', text)

    text=text.replace('\n', ' ')
    text=text.translate(str.maketrans('', '', string.punctuation))
    text=text.strip(' ')
    return text
  else:
        return ""

def casefoldingText(text):
    text=text.lower()
    return text

def tokenizingText(text):
    text=word_tokenize(text)
    return text

def filteringText(text):
    listStopwords = set(stopwords.words('indonesian'))
    filtered=[]
    for txt in text:
      if txt not in listStopwords:
        filtered.append(txt)
    text=filtered
    return text

def stemmingText(text):
    factory = StemmerFactory()
    stemmer = factory.create_stemmer()
    text=[stemmer.stem(word) for word in text]
    return text

def toSentence(list_words):
    sentence = ' '.join(word for word in list_words)
    return sentence
```

```python
def strip_emoji(text):
  emoji_pattern = re.compile("["
  u"\U0001F600-\U0001F64F"  # emoticon
  u"\U0001F300-\U0001F5FF"  # simbol & transportasi
  u"\U0001F680-\U0001F6FF"  # transportasi
  u"\U0001F700-\U0001F77F"  # alat olahraga
  u"\U0001F780-\U0001F7FF"  # bendera
  u"\U0001F800-\U0001F8FF"  # suplemen emoji
  u"\U0001F900-\U0001F9FF"  # emoji wajah
  u"\U0001FA00-\U0001FA6F"  # emoji tangan
  u"\U0001FA70-\U0001FAFF"  # alat olahraga
  u"\U0001F004-\U0001F0CF"  # kartu domino
  u"\U0001F170-\U0001F251"  # simbol
  u"\U0001F300-\U0001F5FF"  # simbol & transportasi
  u"\U0001F600-\U0001F64F"  # emoticon
  u"\U0001F680-\U0001F6FF"  # transportasi
  u"\U0001F700-\U0001F77F"  # alat olahraga"
  u"\U0001F780-\U0001F7FF"  # bendera
  u"\U0001F800-\U0001F8FF"  # suplemen emoji
  u"\U0001F900-\U0001F9FF"  # emoji wajah
  u"\U0001FA00-\U0001FA6F"  # emoji tangan
  u"\U0001FA70-\U0001FAFF"  # alat olahraga
  u"\U0001F004-\U0001F0CF"  # kartu domino
  u"\U0001F170-\U0001F251"  # simbol"
  "]+", flags=re.UNICODE)
  text = emoji_pattern.sub(r"", text)
  return text
```

```python
plays['text_clean']=plays['text'].apply(cleaningText)
plays['text_clean']=plays['text_clean'].apply(casefoldingText)
plays['text_clean']=plays['text_clean'].apply(strip_emoji)
plays.drop(['text'], axis=1,inplace=True)

plays['text_preprocessed']=plays['text_clean'].apply(tokenizingText)
plays['text_preprocessed']=plays['text_preprocessed'].apply(filteringText)
plays['text_preprocessed']=plays['text_preprocessed'].apply(stemmingText)

plays.drop_duplicates(subset=['text_clean'], inplace=True)
```

# **PELABELAN LEXICON**

```python
lexicon_positive=dict()
import csv

with open('drive/MyDrive/final_bpml/positive.csv', mode='r') as csvfile:
    reader = csv.reader(csvfile, delimiter=',')
    for row in reader:
        lexicon_positive[row[0]] = int(row[1])

lexicon_negative=dict()
import csv
with open('drive/MyDrive/final_bpml/negative.csv', mode='r') as csvfile:
    reader = csv.reader(csvfile, delimiter=',')
    for row in reader:
        lexicon_negative[row[0]] = int(row[1])

def sentiment_analysis_lexicon_indonesia(text):
    score = 0
    for word in text:
        if (word in lexicon_positive):
            score = score + lexicon_positive[word]
    for word in text:
        if (word in lexicon_negative):
            score = score + lexicon_negative[word]
    polarity=''
    if (score > 0):
        polarity='positive'
    elif (score < 0):
        polarity='negative'
    else:
        polarity='neutral'
    return score, polarity
```

```python
results = plays['text_preprocessed'].apply(sentiment_analysis_lexicon_indonesia)
results=list(zip(*results))
plays['polarity_score'] = results[0]
plays['polarity']=results[1]
print(plays['polarity'].value_counts())
plays.shape
```

```python
class CountVectorizer:
    def __init__(self, max_features=None):
        self.max_features = max_features
        self.vocabulary = None
        self.vocabulary_mapping = None

    def fit(self, documents):
        word_counts = {}
        for doc in documents:
            words = doc.split()
            for word in words:
                if word in word_counts:
                    word_counts[word] += 1
                else:
                    word_counts[word] = 1

        # Sort dan ambil hanya sejumlah max_features kata pertama
        sorted_words = sorted(word_counts, key=word_counts.get, reverse=True)
        if self.max_features:
            sorted_words = sorted_words[:self.max_features]

        self.vocabulary = sorted_words
        self.vocabulary_mapping = {word: idx for idx, word in enumerate(self.vocabulary)}

    def transform(self, documents):
        if self.vocabulary_mapping is None:
            raise ValueError("Fit the CountVectorizer first before transforming data.")

        X = []
        for doc in documents:
            word_counts = [0] * len(self.vocabulary)
            words = doc.split()
            for word in words:
                if word in self.vocabulary_mapping:
                    word_counts[self.vocabulary_mapping[word]] += 1
            X.append(word_counts)
        return X

    def get_feature_names(self):
        return self.vocabulary
```

# **EKSTRAKSI FITUR TF-IDF**

```python
import math


class Tfidf:
    def fit(self, X):
        self.idf = [0] * len(X[0])
        N = len(X)

        # Menghitung IDF (Inverse Document Frequency) untuk setiap kata
        for i in range(len(X[0])):
            doc_count = sum([1 for doc in X if doc[i] > 0])
            if doc_count > 0:
                self.idf[i] = math.log(N / (doc_count+1))

    def transform(self, X):
        X_tf = []
        for doc in X:
            doc_tf = [tf * idf for tf, idf in zip(doc, self.idf)]
            X_tf.append(doc_tf)
        return X_tf
```

```python
tfidf = Tfidf()
tfidf.fit(X_cv_top)
X_tf = tfidf.transform(X_cv_top)
```

```python
import numpy as np
import pandas as pd

df = pd.read_csv(r'drive/MyDrive/final_bpml/labelling.csv')
df

sentiment_mapping = {'neutral': 0, 'negative': 1, 'positive': 2}
df['polarity'] = df['polarity'].map(sentiment_mapping)

df = df[['text_clean', 'polarity']]
df
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, GRU, Embedding
from tensorflow.keras.optimizers import Adam
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

df['text_clean'] = df['text_clean'].astype(str).fillna('')

tfidf = TfidfVectorizer(max_features=5000)  # Ambil 5000 fitur kata paling penting
X_tfidf = tfidf.fit_transform(df['text_clean']).toarray()

X_train, X_test, y_train, y_test = train_test_split(X_tfidf, df['polarity'], test_size=0.2, random_state=42)

model = Sequential([
    Dense(512, activation='relu', input_shape=(X_train.shape[1],)),
    Dropout(0.3),
    Dense(256, activation='relu'),
    Dropout(0.3),
    Dense(128, activation='relu'),
    Dropout(0.3),
    Dense(len(np.unique(df['polarity'])), activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy', optimizer=Adam(learning_rate=0.001), metrics=['accuracy'])

history = model.fit(X_train, y_train, epochs=10, batch_size=64, validation_data=(X_test, y_test))

loss, accuracy = model.evaluate(X_test, y_test)
print(f"Test Accuracy: {accuracy:.4f}")
```
